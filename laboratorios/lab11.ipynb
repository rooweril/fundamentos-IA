```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
grafo = nx.DiGraph()

grafo.add_weighted_edges_from([
    ("A", "B", 8),
    ("A", "D", 5),
    ("A", "E", 4),
    ("B", "C", 3),
    ("B", "F", 4),
    ("C", "G", 2),
    ("C", "F", 9),
    ("D", "E", 3),
    ("D", "H", 6),
    ("F", "E", 3),
    ("F", "K", 8),
    ("E", "I", 5),
    ("H", "I", 4),
    ("H", "M", 2),
    ("I", "J", 4),
    ("I", "N", 3),
    ("I", "M", 1),
    ("M", "N", 1),
    ("K", "J", 6),
    ("J", "O", 9),
    ("G", "K", 3),
    ("G", "L", 2),
    ("L", "K", 5),
    ("L", "P", 3),
    ("P", "K", 2),
    ("N", "O", 2),
    ("O", "P", 12)
])
```

```python
# Imprimimos el grafo

pos = nx.spring_layout(grafo, seed=12)

nx.draw(
    grafo,
    pos,
    with_labels=True,
    node_size=1500,
    arrows=True
)

labels = nx.get_edge_attributes(grafo, "weight")

nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels=labels
)

plt.show()
```

```python
# ¿Puedo moverme de A a B?
print(grafo.has_edge("A", "B"))
```

```python
# Mostramos las rutas

rutas = list(
    nx.all_simple_paths(
        grafo,
        source="A",
        target="P"
    )
)
```

```python
# Imprimimos las rutas

for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}:{ruta}")
```

```python
# Calculamos el costo

def costo_ruta(grafo, ruta):
    costo = 0

    for i in range(len(ruta)-1):
        costo += grafo[ruta[i]][ruta[i+1]]["weight"]

    return costo
```

```python
# Imprimimos el costo

for i, ruta in enumerate(rutas, start=1):
    costo = costo_ruta(grafo, ruta)

    print(f"Ruta {i}:{ruta} -> {costo} unidades")
```

```python
# Elegir la ruta más corta

costos = [
    costo_ruta(grafo, ruta)
    for ruta in rutas
]

mejor_ruta = costos.index(min(costos))

peor_ruta = costos.index(max(costos))
```

```python
# Imprimir la mejor ruta

print(f"La mejor ruta es: {mejor_ruta+1}")

print(
    f"La ruta es: {rutas[mejor_ruta]}"
)

print(
    f"El camino más corto es: {costos[mejor_ruta]}"
)

print(f"La mejor ruta es: {peor_ruta+1}")

print(
    f"El camino más largo es: {costos[peor_ruta]}"
)
```
