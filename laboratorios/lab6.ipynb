```python
#importar las librerias
import matplotlib.pyplot as plt #librerias para graficar
import networkx as nx
```

```python
#creamos una instancia de la clase  graph
#Graph  es una coleccion de nodos
grafo = nx.Graph()
```

```python
#Creamos el grafo utilizando el metodo add_edges
grafo.add_edges_from(
    [
        ("A","B"),
        ("A","C"),
        ("B","D"),
        ("C","G"),
        ("G","E"),
        ("G","J"),
        ("D","F"),
        ("F","H"),
        ("H","I"),
        ("I","J"),
        ("E","F")
    ]
)
```

```python
#Mostrar el grafo

nx.draw(
    grafo,
    with_labels = True,
    node_size=1000
)
plt.show
```

```python
# Mostrar las rutas posibles del grafo
rutas= list (
    nx.all_simple_paths(grafo, source="A", target="J")
)
```

```python
#imprimir las rutas
print("Rutas encontradas:")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
#mostar el total de pasos dd cada ruta
pasos1= len(rutas[0])-1
print ('pasosde la ruta 1: ', pasos1)
pasos2= len(rutas[1])-1
print ('pasos de la ruta 2: ', pasos2)
pasos3= len(rutas[2])-1
print ('pasos de la ruta 3: ', pasos3)
pasos4= len(rutas[3])-1
print ('pasos de la ruta 4: ', pasos4)
```

```python
#aplicamos la logica proposicional
P = pasos1 < pasos2 and pasos1 < pasos3 and pasos1 < pasos4 # P: la mejor ruta es la 1
Q = pasos2 < pasos1 and pasos2 < pasos3 and pasos2 < pasos4 # P: la mejor ruta es la 2
R = pasos3 < pasos1 and pasos3 < pasos2 and pasos3 < pasos4 # P: la mejor ruta es la 3
S = pasos4 < pasos1 and pasos4 < pasos2 and pasos4 < pasos3 # P: la mejor ruta es la 4


if P==True:
    print ('P = True')
    print ('la ruta 1 es mejor opcion')
elif Q==True:
    print ('Q = True')
    print ('la ruta 2 es mejor opcion')
elif R==True:
    print ('R = True')
    print ('la ruta 3 es mejor opcion')
elif S==True:
    print ('S = True')
    print ('la ruta 4 es mejor opcion')

else:
    print ('P = False')
    print ('Q = False')
    print ('R = False')
    print ('S = False')
    print ('las 4 son buenas, haz lo que quieras')
```
