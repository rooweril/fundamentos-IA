```python
import matplotlib.pyplot as plt
import networkx as nx
```

```python
grafo = nx.Graph()
```

```python
grafo.add_edges_from(
    [
        ("servidor_1", "routerA"),
        ("routerA", "switchA"),
        ("switchA", "PC1"),
        ("switchA", "impresoraA"),
        ("routerA", "switchB"),
        ("switchB", "PC2"),
        ("switchB", "impresoraA")
    ]
)
```

```python
#Mostrar el grafo

nx.draw(
    grafo,
    with_labels = True,
    node_size=1000
)
plt.show
```
